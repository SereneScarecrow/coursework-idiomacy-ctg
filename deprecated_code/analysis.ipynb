```python
import pandas as pd
from pathlib import Path

# Загрузка данных
data1 = pd.read_csv(Path.cwd().parent / "data/all_generated_metrics.csv", index_col=0)

# Группировка и статистика
result = (
    data1.groupby(['model', 'size', 'level'])["idiomatic_density"]
    .agg(['mean', 'std', 'min', 'max', 'count'])
    .round(4)
    .reset_index()
)

result


data1['text_domen'] = data1['index_orig'].map(data.set_index('Unnamed: 0')['text_domen'])
```

```python
def count_text_words(text):

    tokens = str(text).split()

    words_count = sum(1 for token in tokens if any(ch.isalpha() for ch in token))
    
    return words_count

data["len_generated"] = data['text'].apply(count_text_words)
data["len_idioms"] = data['idiomatic_expressions'].apply(count_text_words)
data["idiomatic_density"] = data["len_idioms"] / data["len_generated"]

data.head()

data.to_csv(Path.cwd().parent / "data/all_generated_judge.csv")
```
